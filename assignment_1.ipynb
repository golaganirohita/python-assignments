```python
def celsius_to_fahrenheit(c):
    return (c * 9/5) + 32

def fahrenheit_to_celsius(f):
    return (f - 32) * 5/9

c = float(input("Enter Celsius: "))
print("Fahrenheit =", celsius_to_fahrenheit(c))

f = float(input("Enter Fahrenheit: "))
print("Celsius =", fahrenheit_to_celsius(f))
```

```python
def calculate_grade(avg):
    if avg >= 90:
        return 'A'
    elif avg >= 75:
        return 'B'
    elif avg >= 60:
        return 'C'
    elif avg >= 40:
        return 'D'
    else:
        return 'F'
m1 = int(input("Enter marks 1: "))
m2 = int(input("Enter marks 2: "))
m3 = int(input("Enter marks 3: "))
avg = (m1 + m2 + m3) / 3
print("Average =", avg)
print("Grade =", calculate_grade(avg))
```

```python
def check_even_odd(n):
    if n % 2 == 0:
        print("Even Number")
    else:
        print("Odd Number")
def check_prime(n):
    if n < 2:
        print("Not Prime")
        return
    for i in range(2, n):
        if n % i == 0:
            print("Not Prime")
            return
    print("Prime Number")
num = int(input("Enter a number: "))
check_even_odd(num)
check_prime(num)
```
